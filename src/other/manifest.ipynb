```python
from pathlib import Path
import pandas as pd
import json

PROJECT_ROOT = Path("/home/jovyan/MSC_PROJECT")
DATA_DIR = PROJECT_ROOT / "data" / "avdeepfake1mpp"
METADATA_DIR = DATA_DIR / "metadata"
MANIFEST_DIR = DATA_DIR / "manifests"
EXTRACTED_VAL = DATA_DIR / "extracted" / "val" / "val"

val_meta_path = METADATA_DIR / "val_metadata.json"

with open(val_meta_path, "r") as f:
    data = json.load(f)

if isinstance(data, list):
    meta_df = pd.DataFrame(data)
elif isinstance(data, dict):
    meta_df = pd.DataFrame.from_dict(data, orient="index").reset_index()
    meta_df = meta_df.rename(columns={"index": "metadata_key"})
else:
    raise ValueError(f"Unexpected metadata type: {type(data)}")

print("Metadata shape:", meta_df.shape)
print("Columns:")
print(meta_df.columns.tolist())

display(meta_df.head(20))
```

```python
audio_audit_path = MANIFEST_DIR / "audio_audit_all_videos.csv"
audio_df = pd.read_csv(audio_audit_path)

audio_df["relative_path"] = audio_df["relative_path"].astype(str).map(normalise_path)

keep_audio_cols = [
    "relative_path",
    "has_audio_stream",
    "audio_decodable",
    "audio_codec",
    "sample_rate",
    "channels",
    "audio_duration",
]

keep_audio_cols = [c for c in keep_audio_cols if c in audio_df.columns]

manifest = manifest.merge(
    audio_df[keep_audio_cols],
    on="relative_path",
    how="left"
)

print(manifest.shape)
print("Audio decodable:")
print(manifest["audio_decodable"].value_counts(dropna=False))
```

```python
def labels_from_modify_type(x):
    s = str(x).lower()

    if s == "real":
        return pd.Series({
            "binary_label": 0,
            "visual_fake": 0,
            "audio_fake": 0,
            "condition": "real"
        })

    visual_fake = int("visual" in s or "video" in s or "both" in s)
    audio_fake = int("audio" in s or "both" in s)

    if visual_fake == 1 and audio_fake == 1:
        condition = "fake_video_fake_audio"
    elif visual_fake == 1 and audio_fake == 0:
        condition = "fake_video_real_audio"
    elif visual_fake == 0 and audio_fake == 1:
        condition = "real_video_fake_audio"
    else:
        condition = "fake_unknown"

    return pd.Series({
        "binary_label": 1,
        "visual_fake": visual_fake,
        "audio_fake": audio_fake,
        "condition": condition
    })

label_df = manifest["modify_type"].apply(labels_from_modify_type)
manifest = pd.concat([manifest, label_df], axis=1)

print(manifest["condition"].value_counts(dropna=False))
print(manifest["binary_label"].value_counts(dropna=False))
```

```python
manifest_path = MANIFEST_DIR / "full_val_labelled_manifest.csv"
manifest.to_csv(manifest_path, index=False)

print("Saved:", manifest_path)
print("Rows:", len(manifest))
```
